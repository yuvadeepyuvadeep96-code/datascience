```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips=sns.load_dataset("tips")
```

```python
tips.head()
```

```python
print("Shape of dataset:", tips.shape)
print("Column names:")
print(tips.columns)
print("Data types:")
print(tips.dtypes)
```

```python
print("Missing values:")
print(tips.isnull().sum())
```

```python
categorical_columns = ['sex', 'smoker', 'day', 'time']

for column in categorical_columns:
    tips[column] = tips[column].fillna(tips[column].mode()[0])

print(tips.isnull().sum())
```

```python
categorical_columns = ['sex', 'smoker', 'day', 'time']

print(tips[categorical_columns].head())
```

```python
sex_frequency = tips['sex'].value_counts()

print("Frequency Distribution of Sex:")
print(sex_frequency)
```

```python
smoker_frequency = tips['smoker'].value_counts()

print("Frequency Distribution of Smoker:")
print(smoker_frequency)
```

```python
day_percentage = tips['day'].value_counts(normalize=True) * 100

print("Percentage Distribution of Day:")
print(day_percentage.round(2))
```

```python
time_percentage = tips['time'].value_counts(normalize=True) * 100

print("Percentage Distribution of Time:")
print(time_percentage.round(2))
```

```python
print("Most common Sex:", tips['sex'].value_counts().idxmax())
print("Most common Smoker:", tips['smoker'].value_counts().idxmax())
print("Most common Day:", tips['day'].value_counts().idxmax())
print("Most common Time:", tips['time'].value_counts().idxmax())
```

```python
sns.countplot(data=tips, x='sex')

plt.title("Count Plot of Sex")
plt.xlabel("Sex")
plt.ylabel("Count")

plt.show()
```

```python
sex_counts = tips['sex'].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    sex_counts,
    labels=sex_counts.index,
    autopct='%1.1f%%'
)

plt.title("Percentage Distribution of Sex")

plt.show()
```

```python
day_counts = tips['day'].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(day_counts.index, day_counts.values)

plt.title("Bar Chart of Day")
plt.xlabel("Day")
plt.ylabel("Frequency")

plt.show()
```
