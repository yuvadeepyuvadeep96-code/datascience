```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Mca/html/flights.csv")
```

```python
print("First 5 rows:")
print(df.head())
```

```python
print("\nDataset Shape:")
print(df.shape)
```

```python
print("\nColumn Names:")
print(df.columns)
```

```python
print("\nData Types:")
print(df.dtypes)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df['tailnum'] = df['tailnum'].fillna('Unknown')

print("\nMissing Values After Handling:")
print(df[['carrier', 'tailnum', 'origin', 'dest', 'name']].isnull().sum())
```

```python
categorical_columns = ['carrier', 'origin', 'dest', 'name']

print("\nCategorical Columns:")
print(categorical_columns)
```

```python

for column in categorical_columns:

    print("\n===================================")
    print("Frequency Distribution:", column)
    print("===================================")

    frequency = df[column].value_counts()

    print(frequency)
```

```python

for column in categorical_columns:

    print("\n===================================")
    print("Percentage Distribution:", column)
    print("===================================")

    percentage = df[column].value_counts(normalize=True) * 100

    print(percentage.round(2))
```

```python
for column in categorical_columns:

    most_common = df[column].value_counts().idxmax()
    count = df[column].value_counts().max()

    print("\nColumn:", column)
    print("Most Common Category:", most_common)
    print("Frequency:", count)
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(data=df, x='carrier')

plt.title("Flight Count by Carrier")
plt.xlabel("Carrier")
plt.ylabel("Number of Flights")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='origin')

plt.title("Flight Count by Origin")
plt.xlabel("Origin Airport")
plt.ylabel("Number of Flights")

plt.show()
```

```python
dest_counts = df['dest'].value_counts().head(10)

plt.figure(figsize=(10, 6))

plt.bar(dest_counts.index, dest_counts.values)

plt.title("Top 10 Destinations by Number of Flights")
plt.xlabel("Destination")
plt.ylabel("Number of Flights")

plt.xticks(rotation=45)

plt.show()
```

```python
origin_counts = df['origin'].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    origin_counts.values,
    labels=origin_counts.index,
    autopct='%1.1f%%'
)

plt.title("Flight Distribution by Origin")

plt.show()
```

```python
carrier_counts = df['carrier'].value_counts()

plt.figure(figsize=(9, 9))

plt.pie(
    carrier_counts.values,
    labels=carrier_counts.index,
    autopct='%1.1f%%'
)

plt.title("Flight Distribution by Carrier")

plt.show()
```
