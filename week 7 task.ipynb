```python
print(abc.isnull().sum())
```

```python
abc.fillna({
    col:abc[col].mode()[0]
    for col in abc.columns
    if abc[col].isnull().any()
},inplace=True)
print(abc.isnull().sum())
```

```python
abc.drop_duplicates(inplace=True)
if "Date" in abc.columns:
    abc["Date"] = pd.to_datetime(abc["Date"], errors="coerce")

print(abc.dtypes)
print(abc.head())
```

```python
print(abc.loc[0:4, :])
print(abc.loc[:, ["Crop", "Production"]])
print(abc.iloc[0:5, :])
print(abc.iloc[0:10, 0:5])
```

```python
print(abc[abc["Production"] > 1000])
print(abc[abc["Production"] < 1000])
print(abc[abc["Production"] >= 1000])
print(abc[(abc["Production"] > 1000) &
         (abc["Crop"] == "Rice")])
print(abc[(abc["Production"] > 1000) |
         (abc["Crop"] == "Rice")])
print(abc[~(abc["Crop"] == "Rice")])
```

```python
abc["Production_in_Thousands"] = abc["Production"] / 1000
print(abc.head())
```

```python
print(abc.groupby("Crop")["Production"].sum())
print(abc.groupby("Crop")["Production"].mean())
print(abc.groupby("Crop")["Production"].agg(
    ["min", "max", "sum", "mean", "count", "std"]
))
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

abc["Crop_Encoded"] = le.fit_transform(abc["Crop"])

print(abc[["Crop", "Crop_Encoded"]].head())
```

```python
abc_onehot = pd.get_dummies(
    abc,
    columns=["Season"],
    dtype=int
)

print(abc.head())
```

```python
print(abc.columns.tolist())
```

```python
print(abc["Season"].unique())
```

```python
from sklearn.preprocessing import OrdinalEncoder
abc["Season"] = abc["Season"].str.strip()
order = [[
    "Autumn",
    "Kharif",
    "Rabi",
    "Summer",
    "Winter",
    "Whole Year"
]]
oe = OrdinalEncoder(categories=order)
abc["Season_Encoded"] = oe.fit_transform(
    abc[["Season"]]
)
print(abc[["Season", "Season_Encoded"]].head())
```

```python
abc = pd.to_numeric(abc["Production"], errors="coerce").dropna()

print(abc.head())
```

```python
print("Mean:", abc.mean())
print("Median:", abc.median())
print("Mode:", abc.mode().tolist())
```

```python
print("Standard Deviation:", abc.std())
print("Variance:", abc.var())
```

```python
minimum = abc.min()
maximum = abc.max()
range_value = maximum - minimum

print("Minimum:", minimum)
print("Maximum:", maximum)
print("Range:", range_value)
```

```python
Q1 = abc.quantile(0.25)
Q2 = abc.quantile(0.50)
Q3 = abc.quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
print("Skewness:", abc.skew())
```

```python
print("Kurtosis:", abc.kurt())
```

```python
Q1 = abc.quantile(0.25)
Q3 = abc.quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = abc[
    (abc < lower_limit) |
    (abc > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:")
print(outliers)
print("Number of Outliers:", len(outliers))
```

```python
mean_value = abc.mean()
std_value = abc.std()

if mean_value != 0:
    CV = (std_value / abs(mean_value)) * 100
    print("Coefficient of Variation:", CV, "%")
else:
    print("CV cannot be calculated because mean is zero.")
```

```python
mean_value = abc.mean()
median_value = abc.median()

print("Mean:", mean_value)
print("Median:", median_value)

if mean_value > median_value:
    print("Distribution is likely positively skewed.")
elif mean_value < median_value:
    print("Distribution is likely negatively skewed.")
else:
    print("Mean and median are equal; distribution may be symmetric.")
```
